# scitex.plt -- Reproducible, Restylable Figures

Powered by figrecipe. Data and style are decomposed: figures can be restyled without altering the data hash.

```python
%matplotlib inline
import scitex as stx
import numpy as np

np.random.seed(42)
```

## Basic Line Plot

```python
fig, ax = stx.plt.subplots()

x = np.linspace(0, 10, 200)
ax.stx_line(x, np.sin(x))
ax.set_xyt("Time", "Amplitude", "Sine")

stx.io.save(fig, "/tmp/line.png")
fig
```

## Multi-Panel Figure

```python
fig, axes = stx.plt.subplots(1, 3)

# Panel 1: Line plot
x = np.linspace(0, 2 * np.pi, 100)
axes[0].stx_line(x, np.cos(x))
axes[0].set_xyt("x", "y", "Cosine")

# Panel 2: Violin plot
data = [np.random.normal(0, 1, 200), np.random.normal(1, 1.5, 200)]
axes[1].stx_violin(data)
axes[1].set_xyt("Group", "Value", "Violin")

# Panel 3: Heatmap
mat = np.random.rand(8, 8)
axes[2].stx_heatmap(mat)
axes[2].set_xyt("Col", "Row", "Heatmap")

stx.io.save(fig, "/tmp/multi_panel.png")
fig
```

## Statistical Visualization

```python
fig, axes = stx.plt.subplots(1, 2)

# Mean +/- SD
data = np.random.normal(loc=[2, 4, 6], scale=[0.5, 1.0, 0.8], size=(50, 3))
axes[0].stx_mean_std(data)
axes[0].set_xyt("Condition", "Value", "Mean +/- SD")

# Median +/- IQR
axes[1].stx_median_iqr(data)
axes[1].set_xyt("Condition", "Value", "Median +/- IQR")

stx.io.save(fig, "/tmp/stat_viz.png")
fig
```

## Available Plot Types

```python
# Key stx.plt plot types (all produce reproducible YAML recipes)
plot_types = [
    "stx_line",          # Line plot with optional CI shading
    "stx_violin",        # Violin plot
    "stx_heatmap",       # Annotated heatmap
    "stx_scatter_hist",  # Scatter with marginal histograms
    "stx_conf_mat",      # Confusion matrix
    "stx_raster",        # Raster / event plot
    "stx_mean_std",      # Mean +/- standard deviation
    "stx_mean_ci",       # Mean +/- confidence interval
    "stx_median_iqr",    # Median +/- interquartile range
    "stx_shaded_line",   # Line with shaded error band
    "stx_ecdf",          # Empirical CDF
    "stx_fillv",         # Vertical fill bands
    "stx_rectangle",     # Rectangle annotations
    "stx_image",         # Image display
]

for pt in plot_types:
    print(f"  ax.{pt}()")
```

## Auto CSV Export

When you call `stx.io.save(fig, "plot.png")`, scitex automatically:

1. Saves the image (PNG, PDF, SVG, etc.)
2. Exports a **YAML recipe** (`plot.yaml`) capturing data + style
3. Exports **CSV files** (`plot_data/*.csv`) for each subplot's data

This means every figure is reproducible and the underlying data is always accessible for review or re-analysis.
