# scitex.scholar -- Literature Management

Search, download, and enrich academic papers. Backed by local CrossRef (167M+) and OpenAlex (250M+) databases.

````python
import scitex as stx
````

## Search for Papers

````python
results = stx.scholar.search("neural oscillations working memory", limit=5)
````

## Fetch Paper by DOI

````python
stx.scholar.fetch("10.1038/s41586-024-07804-3")
````

## Enrich BibTeX

````python
stx.scholar.enrich_bibtex("references.bib", output="enriched.bib")
````

## Parse BibTeX

````python
entries = stx.scholar.parse_bibtex("references.bib")
````

## CLI Equivalents

```bash
scitex scholar search 'neural oscillations' --limit 20
scitex scholar fetch '10.1038/s41586-024-07804-3'
scitex scholar bibtex references.bib
```
