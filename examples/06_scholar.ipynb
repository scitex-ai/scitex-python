# scitex.scholar -- Literature Management

Search, download, and enrich academic papers. Backed by local CrossRef (167M+) and OpenAlex (250M+) databases.

````python
import scitex as stx
````

## Search for Papers

````python
# Search for papers (requires CrossRef/OpenAlex database or internet)
try:
    results = stx.scholar.search("neural oscillations working memory", limit=5)
    for r in results[:3]:
        print(f"  {r.get('title', 'N/A')[:80]}")
except Exception as e:
    print(f"Scholar search requires database setup: {type(e).__name__}")
````

## Fetch Paper by DOI

````python
# Fetch paper by DOI (requires browser/network)
# stx.scholar.fetch("10.1038/s41586-024-07804-3")
print("stx.scholar.fetch(doi) downloads PDF via browser-based authentication")
````
