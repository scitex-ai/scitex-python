# scitex.writer -- LaTeX Manuscript Compilation

```python
import scitex as stx
```

## Compile Manuscript

```python
stx.writer.compile_manuscript("paper/")
```

## Add Figures and Tables

```python
stx.writer.add_figure(
    project_dir="paper/",
    label="fig:accuracy",
    caption="Classification accuracy across conditions.",
    path="figures/accuracy.pdf",
)

stx.writer.add_table(
    project_dir="paper/",
    label="tab:demographics",
    caption="Participant demographics.",
    path="tables/demographics.csv",
)
```

## Bibliography Management

```python
stx.writer.list_bibentries("paper/")

stx.writer.add_bibentry(
    project_dir="paper/",
    doi="10.1038/s41586-024-07804-3",
)
```

## Claims (linked to Clew verification)

```python
stx.writer.add_claim(
    project_dir="paper/",
    claim_type="statistical",
    text="Classification accuracy was significantly above chance (p < 0.001).",
    source_session="./sessions/2026-03-15_exp01/",
)
```
