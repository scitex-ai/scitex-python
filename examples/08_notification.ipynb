# scitex.notification -- Multi-Backend Notifications

Get notified when experiments finish via desktop, phone call, SMS, or email with automatic fallback.

````python
import scitex as stx
````

## Alert (with automatic backend fallback)

````python
# Alert with automatic backend fallback (desktop -> email -> SMS)
try:
    stx.notification.alert("Experiment complete: accuracy = 94.2%")
except Exception as e:
    print(f"alert(): {type(e).__name__} (configure backends in .env.d/)")
````

## Phone Call (for critical events)

````python
# Phone call for critical events (requires Twilio credentials)
# stx.notification.call("Training diverged -- loss is NaN")
print("stx.notification.call(msg) requires TWILIO_ACCOUNT_SID, TWILIO_AUTH_TOKEN")
````

## SMS

````python
# SMS notification (requires Twilio credentials)
# stx.notification.sms("GPU job finished on node-42")
print("stx.notification.sms(msg) requires TWILIO_ACCOUNT_SID, TWILIO_AUTH_TOKEN")
````

## Available Backends

````python
# List available notification backends
backends = stx.notification.available_backends()
print("Available backends:", backends)
````
