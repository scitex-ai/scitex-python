## LLM-Friendly Types

```python
from scitex.dev import Result, supports_return_as

# Result wraps success/failure with structured metadata
ok = Result(success=True, data={"version": "0.5.0"})
print("Success:", ok)

err = Result(success=False, error="Package not installed", error_code="NOT_FOUND")
print("Error:", err)


# @supports_return_as adds return_as="result" option to any function
@supports_return_as
def my_tool(x, y):
    return x + y


# Normal call -- returns raw value
print("Normal:", my_tool(1, 2))

# With return_as="result" -- wraps in Result object
print("Result:", my_tool(1, 2, return_as="result"))
```

## Wrap Functions as MCP Tools

```python
def my_function(name: str, count: int = 5) -> dict:
    """Greet someone multiple times."""
    return {"greeting": f"Hello, {name}!", "count": count}


# Generate an MCP tool definition from any typed function
tool_def = stx.dev.wrap_as_mcp(my_function)
print(tool_def)
```
