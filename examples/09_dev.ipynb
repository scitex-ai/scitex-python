## LLM-Friendly Types

```python
from scitex.dev import Result, ErrorCode, supports_return_as

# Result wraps success/failure with structured metadata
ok = Result.ok({"version": "0.5.0"})
print(ok)

err = Result.err(ErrorCode.NOT_FOUND, "Package not installed")
print(err)


# supports_return_as marks functions that can return JSON, text, or dict
@supports_return_as
def my_tool():
    return Result.ok({"status": "healthy"})


print(my_tool(return_as="json"))
```

## Wrap Functions as MCP Tools

```python
def my_function(name: str, count: int = 5) -> dict:
    """Greet someone multiple times."""
    return {"greeting": f"Hello, {name}!", "count": count}


# Generate an MCP tool definition from any typed function
tool_def = stx.dev.wrap_as_mcp(my_function)
print(tool_def)
```
